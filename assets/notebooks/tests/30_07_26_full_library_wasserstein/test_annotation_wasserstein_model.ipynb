```python
# Creating wrapper 

## imports 
from msi_autoencoder_wrapper import MSIAutoEncoderWrapper

## setting workspace
workspace_path = 'workspace'

wrapper = MSIAutoEncoderWrapper(
    project_path='workspace',
)
```

```python
# Reader setup
wrapper.workspace.set_default_image_path('kidney-pilot-v2')
wrapper.context_manager.set_reader('PyImzMLReader') 
wrapper.context_manager.set_binner('LinearBinning', bin_step= 0.01)
wrapper.context_manager.set_inverse_binner('TopPeaksInverseBinner')
```

```python
# obtain annotation information\
print('Metadata', wrapper.active_context.annotation_reader.get_dataset_metadata())
print('Spectra:', wrapper.active_context.reader.GetNumberOfSpectra())
print('Molecular annotations:', len(wrapper.active_context.annotation_reader.get_annotations()))
print('Annotated spectrum 449 length:', len(wrapper.active_context.annotation_reader.get_spectrum_annotations(449)))
print('Annotated spectrum 449:', wrapper.active_context.annotation_reader.get_spectrum_annotations(449))
```

```python
# import helpers for creating annotation
from msi_autoencoder_wrapper.models.datasets.class_assignment import (
    build_class_mapping, molecule_key,
)

# create annotation
##  here we iterate through all filter we have so it contains 
molecule_mapping = build_class_mapping(
    molecule_key(item) for item in wrapper.active_context.annotation_reader.get_annotations()
)

## Here we iterate through names to obtain only names (for print purpose only)
molecule_labels = [None] * len(molecule_mapping)
for label, class_index in molecule_mapping.items():
    molecule_labels[class_index] = label

print('Molecular classes:', len(molecule_mapping))
molecule_mapping
```

Model creation

```python
wrapper.device
```

```python
latent_dim = 128
model_name = 'kidney-annotation-autoencoder_optimization_test'

wrapper.models_manager.set_model_type('autoencoder', model_name)
wrapper.models_manager.set_model_preset(
    'GradualReduction', latent_dim=latent_dim, projection_dim=64
)
wrapper.models_manager.set_head(
    head_id='molecule_primary',
    target_field='molecule',
    strategy='LinearClassificationHead',
    latent_dim=latent_dim,
    output_dim=len(molecule_mapping), # 22 classes for this example
    hidden_dim=64,
    dropout=0.1,
)
wrapper.models_manager.set_dataset(
    'PixelDataset',
    normalization='tic',
    target_specs={
        'molecule': {
            'type': 'multi_label',
            'class_mapping': molecule_mapping,
        },
    },
)
model = wrapper.models_manager.compile_model(run_validation_pass=True)
model
```

```python
# 

## import 
import torch 

## example run to see model dimensions
dataset = wrapper.active_dataset
spectrum_id, spectrum, targets, target_masks = dataset[449]
with torch.no_grad():
    outputs = model(spectrum.unsqueeze(0).to(wrapper.device))
print('Dataset size:', len(dataset))
print('Spectrum shape:', tuple(spectrum.shape))
print('Positive molecule targets:', int(targets['molecule'].sum().item()))
print('Latent shape:', tuple(outputs['latent_space'].shape))
print('Molecule logits shape:', tuple(outputs['head_molecule_primary'].shape))

## check how many pixel are without molecular label
#TODO - during merge tehre should be opotion for "how many unlabbelled molecules" - be default we should take only annoated pixel (write new imzML and ibd file so it would be clear), for background information we should include unannotataed data. it is imprtant as it is says which label are unresgisterd and how to encode "noise" to retrive similiar values.  
target_counts = torch.stack([dataset[index][2]['molecule'] for index in range(len(dataset))]).sum(0)
print('Pixels without a molecular label:', sum(dataset[index][2]['molecule'].sum() == 0 for index in range(len(dataset))))
print('Positive pixels per class:')
for label, count in zip(molecule_labels, target_counts.tolist()):
    print(f'{count:6.0f}  {label}')
```

```python
training_config = {
    'seed': 1912,
    'checkpoint': {'enabled': True, 'restore_best': True},
    'phases': [
        {
            'phase_name': 'reconstruction_and_molecule_classification',
            'epochs': 10,
            'batch_size': 16,
            'dataloader': {'num_workers': 0, 'pin_memory': wrapper.device == 'cuda'},
            'gradient_clip_norm': 1.0,
            'freeze': [],
            'optimizer': {
                'type': 'AdamW',
                'params': {'lr': 1e-3, 'weight_decay': 1e-4},
            },
            'criterions': {
                'reconstruction': {
                    # 'mse': {'target': 'MSELoss', 'weight': 1.0, 'params': {}}, # to na ten moment pomijamy 
                    'masserstein': {
                        'target': 'MassersteinLoss',
                        'weight': 1.0,
                        'params': {'sinkhorn_iterations': 20},
                    },
                },
                'heads': {
                    'molecule_primary': {
                        'molecule_bce': {
                            'target': 'MultiLabelBCELoss',
                            'weight': 0.25,
                            'params': {},
                        },
                    },
                },
            },
        },
    ],
}
training_config
```

```python
## apply train setup 
resource_report = wrapper.models_manager.estimate_training_resources(
    training_config,
    resource_limits={'ram': 0.65, 'vram': 0.80, 'disk': 1.00},
    auto_adjust_batch_size=True,
)
safe_training_config = resource_report['recommended_training_config']
resource_report
```

```python
history = wrapper.models_manager.fit(safe_training_config)
```

Saving results

```python

latent_path = wrapper.active_context.save_latent(
    loader_config={'batch_size': 64, 'num_workers': 0, 'pin_memory': wrapper.device == 'cuda'},
    activate=False,
)
print('Latent dataset:', latent_path)
```

```python
latents = wrapper.models_manager.autoencoder.transform(
    {'batch_size': 64, 'num_workers': 0, 'pin_memory': wrapper.device == 'cuda'}
)
with torch.no_grad():
    logits = model.heads['molecule_primary'](torch.from_numpy(latents).float().to(wrapper.device))
    probabilities = torch.sigmoid(logits).cpu().numpy()
print('Latents:', latents.shape)
print('Molecule probabilities:', probabilities.shape)
print('Finite:', np.isfinite(latents).all() and np.isfinite(probabilities).all())
```

Calculating predictions 

```python
spectrum_id = 449
top_indices = np.argsort(probabilities[spectrum_id])[::-1][:5]
predictions = [
    {
        'molecule': molecule_labels[index],
        'probability': float(probabilities[spectrum_id, index]),
        'target': int(dataset[spectrum_id][2]['molecule'][index].item()),
    }
    for index in top_indices
]
predictions
```
