# 5. Multi-image models — TODO

This tutorial is intentionally a placeholder. The current stable contract covers one active image context and one currently loaded model. A loaded model may be stored under the `global` key, but that storage scope is not yet the multi-image training API.

## Planned scope

The future implementation and tutorial will cover:

- datasets spanning multiple independent image contexts;
- per-image reader, binner, coordinate, normalization, and latent-source configuration;
- sampling and balancing pixels across images;
- global model context, checkpoints, and reproducible multi-image manifests;
- training resource estimates that include all source images and reader caches;
- inference and latent export back to each source image's spatial metadata.

Do not infer this functionality from `workspace.set_active_images(...)` alone. Until the dataset and persistence contracts are implemented and tested, train the documented `PixelDataset` workflow on one selected image at a time.
